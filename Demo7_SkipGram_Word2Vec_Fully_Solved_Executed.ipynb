## Q1 – Skip-Gram Pair Generation

```

sentence = "I love Deep Learning Because Deep Learning is powerful".split()

target_index = 2
target = sentence[target_index]
window_size = 2

pairs = []

for i in range(max(0, target_index-window_size),
               min(len(sentence), target_index+window_size+1)):
    if i != target_index:
        pairs.append((target, sentence[i], 1))

pairs
```

## Q2 – Negative Sampling

```

negative_samples = [("Deep", "powerful", 0),
                    ("Deep", "is", 0)]

all_pairs = pairs + negative_samples
all_pairs
```

## Q3 – Probability Computation

```

import math

deep = [0.3, -0.5]
learning = [0.6, 0.2]
powerful = [-0.4, 0.8]

dot_positive = deep[0]*learning[0] + deep[1]*learning[1]
p_positive = 1/(1+math.exp(-dot_positive))

dot_negative = deep[0]*powerful[0] + deep[1]*powerful[1]
p_negative = 1/(1+math.exp(dot_negative))

print("Positive Dot Product:", round(dot_positive,4))
print("P(+|Deep,Learning):", round(p_positive,4))

print("\nNegative Dot Product:", round(dot_negative,4))
print("P(-|Deep,powerful):", round(p_negative,4))
```

## Q4 – Forward Propagation

```

import numpy as np

W = np.array([
    [0.1, 0.4],
    [0.3, -0.2],
    [0.5, 0.6]
])

one_hot_love = np.array([0,1,0])

love_embedding = one_hot_love @ W
deep_embedding = np.array([0.5,0.6])

z = np.dot(deep_embedding, love_embedding)

print("Embedding of love:", love_embedding)
print("Predicted score z:", round(z,4))
```

## Q5 – Loss & Backpropagation

```

target = 1

loss = 0.5*((target-z)**2)
gradient = z - target

grad_w = gradient * love_embedding

print("Loss:", round(loss,4))
print("Gradient:", round(gradient,4))
print("Weight Gradient:", grad_w)
```

## Q6 – Weight Update

```

alpha = 0.1

updated_love = love_embedding - alpha*grad_w

print("Updated embedding for love:")
print(updated_love)
```
